# Stage 5 Hybrid Reporting
This notebook reads one completed, step-consistent Stage 5 snapshot. It does not evaluate factors or run Stage 6/OOS.

## 00 Parameters and Snapshot Validation

```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == "notebooks":
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

import matplotlib.pyplot as plt
from IPython.display import display

from factor_gfn.gfn import build_stage5_hybrid_variance_5_15_config
from factor_gfn.reporting import Stage5ReportRenderer, load_stage5_report_data
```

```python
REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == 'notebooks':
    REPO_ROOT = REPO_ROOT.parent
RUN_DIR = REPO_ROOT / 'runs' / 'stage5_hybrid_variance_real_5_15' / 'hybrid_5_15_k16_seed42_20260816T025559Z'
# 如需查看其他已完成 run，可改为：Path(r'REPLACE_WITH_COMPLETED_STAGE5_RUN_DIRECTORY')
OUTPUT_DIR = REPO_ROOT / 'outputs' / 'stage5_reporting_v2'
ALLOW_INCOMPLETE = False
EXPECTED_CONFIG = build_stage5_hybrid_variance_5_15_config(max_cycles=100, trajectories_per_batch=16, seed=42)
```

```python
bundle = load_stage5_report_data(
    RUN_DIR,
    expected_config=EXPECTED_CONFIG,
    allow_incomplete=ALLOW_INCOMPLETE,
    correlation_top_k=20,
)
renderer = Stage5ReportRenderer(bundle, OUTPUT_DIR)
display(bundle.snapshot_manifest)
```
