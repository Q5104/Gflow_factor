# Prepare Daily-Derived v1 data

Manual-only entry for the frozen `daily_derived_v1` artifact. This Notebook does not change Raw processed artifacts and the builder refuses to overwrite an existing Derived tensor/metadata.

```python
import sys
from pathlib import Path

working_dir = Path.cwd().resolve()
PROJECT_ROOT = next(
    (path for path in (working_dir, *working_dir.parents) if (path / 'factor_gfn').is_dir()),
    None,
)
if PROJECT_ROOT is None:
    raise RuntimeError('无法从当前工作目录向上找到包含 factor_gfn/ 的项目根目录')
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from factor_gfn.data import (
    DailyDerivedArtifactConfig,
    build_daily_derived_artifact,
    inspect_daily_derived_artifact,
    inspect_daily_derived_inputs,
)

print('项目根目录:', PROJECT_ROOT)
config = DailyDerivedArtifactConfig()
config
```

```python
# Read-only preflight: authority axes, shared universe, and reused PIT shares.
inspect_daily_derived_inputs(config)
```

```python
# Manual full build. Expected output: data/processed/daily_derived_v1/.
RUN_FULL_BUILD = False
if not RUN_FULL_BUILD:
    raise RuntimeError('Safety stop：检查只读 preflight 后再显式设置 RUN_FULL_BUILD=True')
metadata = build_daily_derived_artifact(config)
metadata['shape'], metadata['builder_schema_fingerprint']
```

```python
# Read-only completed-artifact validation and compact QA.
completed = inspect_daily_derived_artifact(config)
completed['qa']
```
